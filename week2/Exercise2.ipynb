```python
#program that find maximum among three number
a=int(input("enter first number"))
b=int(input("enter second number"))
c=int(input("enter third number"))
if(a>b and a>c):
  print(f"(a) is greater")
elif(b>a and b>c):
  print(f"(b) is greater")
else:
  print(f"(c) is greater")
```

```python
#given year is leapyearor not
year=int(input("enter year"))
if(year%4==0 and year%100!=0) or (year%400==0):
  print(f"(year) is a leap year")
else:
  print(f"(year) is not a leap year")
```

```python
#example code on match-case
day=int(input("enter a number"))
match day:
  case 1:
    print("monday")
  case 2:
    print("tuesday")
  case 3:
    print("wednesday")
  case 4:
    print("thursday")
  case 5:
    print("friday")
  case 6:
    print("saturday")
  case 7:
    print("sunday")
  case _:
    print("unvalid input")
```

loops in python

```python
#printing 1 to 10
i=1
while(i<=10):
  print(i,end=" ")
  i=i+1
```

```python
#printing 1 to n even numbers only
n=int(input("enter value where to start print"))
while(n>=1):
  print(n,end=" ")
  n-=1
```

```python
#printing 1 to n even numbers only
n=int(input("enter range"))
i=1
while(i<=n):
  if(i%2==0):
    print(i,end=" ")
  i+=1
```

for loop

```python
#printing 1 to 10
for i in range(1,10):
  print(i,end=" ")
```

```python
#printing n to 1 numbers
n=int(input("enter value where to start print"))
for i in range (n,0,-1):
  print(i,end=" ")
```

```python
#printing 1 to n even numbers
n=int(input("enter range"))
for i in range(0,n,2):
  print(i,end=" ")
```

```python
#program that prints table upto 10 multiple
n=int(input("enter a number"))
for i in range(1,11):
  print(f"{n} * {i} = {n*i}")
```

```python
#program that finds the sum of thr digits of the given number
n=int (input("enter a number"))
sum=0
while(n>0):
   r=n%10
   sum=sum+r
   n=n//10
   print(f"sum of digits in number is {sum}")
```

```python
#to check if the number is prime number or not
n=int(input("enter a number"))
flag=0
for i in range(1,n/2):
  if(n%i==0):
    flag=1
    break
if(flag==0):
    print(f"{n} is a prime number")
else:
    print(f"{n} is not a prime number")
```
